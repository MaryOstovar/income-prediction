## 1. Business Understanding

### 1.1 Problem Statement

Many organizations need to understand the income level of their customers or employees. However, direct income information is often unavailable or considered private, so it cannot always be used. Because of this, organizations sometimes need other ways to estimate a person’s income level.

In this project, the goal is to predict whether a person earns more than $50K per year . The prediction is based on demographic and work-related information from the Adult Census dataset, such as age, education level, occupation, and working hours per week.

To keep the project organized and systematic, it follows the CRISP-DM methodology. This framework helps structure the process from understanding the problem and preparing the data to training models and evaluating the results.

### 2.2 Article 2 Analysis: Reduced, Weighted SVM (RW-SVM) for Income Prediction

I read this paper about a method called Reduced, Weighted SVM (RW-SVM). The idea is interesting—they first use clustering to summarize the training data into centroids, then assign weights to these centroids before training an SVM. This reduces the number of training samples significantly, which makes training faster and uses less memory, while keeping accuracy high. They tested it on the UCI Adult Census dataset for income prediction and also for gender classification.

This method seems particularly useful for imbalanced datasets. By giving more weight to centroids representing minority classes, the model improves predictions for the underrepresented group. Overall accuracy is similar to standard SVM, but minority class performance is noticeably better.

Preprocessing in this study is straightforward: normalize features to [-1, 1], replace missing values with the mean, and convert categorical values to numeric. This aligns well with the steps I plan to use for my dataset. The clustering component could be challenging to implement directly, but the overall concept of making SVM more efficient is informative.

Some limitations I noticed: they only tested binary classification, and their second weighting method does not address class imbalance. Also, the choice of K in clustering may affect results, and they do not discuss feature selection beyond scaling and mapping categorical features.

For my project, the main takeaways are:
- If I use SVM, it is helpful to think about ways to reduce training size to speed up experiments, even if I don’t implement RW-SVM exactly.
- Handling class imbalance is important; giving more importance to the minority class can improve predictions for adults earning >50K.
- Preprocessing steps like normalization and encoding categorical variables are essential before feeding data into SVM.
- I should report both overall accuracy and minority class performance (such as F1-score) to ensure fairness.

Overall, this paper provides useful ideas for improving SVM efficiency and fairness for imbalanced datasets. Even if I do not follow their method exactly, I can apply these insights to my own SVM experiments.

### 2.3 Article 3 Analysis: Income Prediction Using Logistic Regression, Random Forest and SVM

In this paper, the authors study the problem of predicting adult income using the Adult Income dataset. The dataset contains demographic information such as age, education level, working hours, and marital status. They compare three common machine learning models: Logistic Regression, Random Forest, and Support Vector Machine (SVM). The goal is to predict whether a person earns more than $50K per year or not.

From the results, all three models perform quite well. SVM achieved the highest accuracy (about 85.88%), while Logistic Regression achieved the best AUC score. This suggests that even relatively simple models can perform well on this dataset. Another interesting part of the study is the feature importance analysis, which shows that marital status, age, and years of education are among the most important factors influencing income.

The authors split the dataset into 70% for training and 30% for testing. They also used 5-fold cross-validation during training to make the results more reliable. Logistic Regression was used as a baseline model, Random Forest was used to capture more complex relationships in the data, and SVM was applied with an RBF kernel to model nonlinear patterns.

One limitation I noticed is that the paper does not explain the preprocessing steps in detail. For example, it is not very clear how missing values or categorical variables were handled. Also, the study mainly relies on Random Forest for feature importance, without comparing it with other methods.

For my project, this paper gives several useful ideas. First, Logistic Regression can work well as a baseline model because it is simple and easy to interpret. Second, it is important to evaluate models using multiple metrics such as accuracy, precision, recall, F1-score, and AUC, instead of relying on only one metric. Finally, the paper highlights some features that are strongly related to income prediction, such as marital status, age, education level, capital gain, and working hours. These variables will probably be important when I explore the dataset and perform feature analysis.

Overall, this paper shows that even relatively simple machine learning models can achieve strong performance on the Adult Income dataset. It also confirms that demographic and socioeconomic factors play an important role in predicting income levels.
