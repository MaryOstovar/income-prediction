## 1. Business Understanding

### 1.1 Problem Statement

Many organizations need to understand the income level of their customers or employees. However, direct income information is often unavailable or considered private, so it cannot always be used. Because of this, organizations sometimes need other ways to estimate a person’s income level.

In this project, the goal is to predict whether a person earns more than $50K per year . The prediction is based on demographic and work-related information from the Adult Census dataset, such as age, education level, occupation, and working hours per week.

To keep the project organized and systematic, it follows the CRISP-DM methodology. This framework helps structure the process from understanding the problem and preparing the data to training models and evaluating the results.

### 1.2 Business Objectives
The main objectives of my project are:

- Develop a machine learning approach to predict whether a person’s income exceeds $50K using demographic and employment-related attributes.
- Understand how factors like education, occupation, age, and working hours relate to income.
- Provide interpretable insights into which variables are most strongly associated with higher or lower income categories.
- Create a reproducible and structured workflow following the CRISP-DM framework up to the evaluation phase.

I also realized that this project could have practical value. For example, a financial institution could use a similar approach for preliminary income assessments without accessing sensitive financial records.

## 2 Related Work

### 2.1 Article 1 Analysis: Adult Income Prediction Using Multiple Machine Learning Models

This study investigates the problem of predicting individual income levels using the UCI Adult Census dataset. The authors evaluate a wide range of machine learning models, including Logistic Regression, Decision Trees, Random Forest, Support Vector Machines (SVM), XGBoost, and neural network–based ensemble approaches.

The methodology involves several preprocessing steps such as one-hot encoding of categorical variables, scaling of numerical attributes, and the removal of highly correlated or low-impact features. These preprocessing decisions play a crucial role in improving model performance and ensuring that algorithms can effectively learn from mixed data types.

The results show that ensemble approaches combining neural networks and gradient boosting techniques achieve the highest predictive accuracy, reaching approximately **87%**. However, traditional machine learning models such as Random Forest and SVM also achieve competitive performance, demonstrating that strong predictive results can be obtained without deep learning techniques.

Feature importance analysis in the study highlights **education level, occupation, and marital status** as key predictors of income. These findings align with common observations in socioeconomic data analysis and reinforce the importance of demographic and occupational characteristics in income prediction tasks.

Despite its strengths, the study presents some limitations. The authors remove records with missing values rather than applying imputation methods, which may reduce the available training data. Additionally, while fairness considerations are briefly mentioned, the paper does not provide a detailed bias or fairness analysis.

For the current project, this study provides several important insights. First, traditional machine learning models such as **Random Forest, SVM, and Logistic Regression** remain strong candidates when deep learning approaches are not permitted. Second, careful preprocessing—including categorical encoding, feature selection, and scaling—is essential for obtaining reliable results. Finally, the study emphasizes the importance of evaluating models using metrics such as **Precision, Recall, F1-score, and ROC-AUC**, particularly in the presence of class imbalance.

### 2.2 Article 2 Analysis: Reduced Weighted Support Vector Machine (RW-SVM)

This paper introduces a modified Support Vector Machine framework called **Reduced Weighted SVM (RW-SVM)**, designed to improve computational efficiency while maintaining predictive performance. The approach first applies clustering to the training data in order to generate representative centroids. These centroids are then assigned weights and used as the training instances for the SVM model.

The main advantage of this approach is the reduction of the effective training dataset size. By summarizing the original dataset through cluster centroids, the algorithm reduces computational complexity, training time, and memory usage while preserving the most informative patterns in the data.

The authors evaluate the method using the **UCI Adult Census dataset** for both income prediction and gender classification tasks. Experimental results show that RW-SVM achieves performance comparable to standard SVM models while significantly reducing training costs. In addition, the weighting mechanism improves prediction quality for minority classes, which is particularly beneficial when working with imbalanced datasets.

The preprocessing pipeline used in the study includes **feature normalization to the range [-1,1]**, replacement of missing values with mean values, and conversion of categorical variables into numerical representations. These preprocessing steps ensure compatibility with the SVM algorithm and improve model stability.

However, several limitations are identified in the study. The experiments focus exclusively on binary classification tasks, and the influence of clustering parameters—such as the number of clusters—on model performance is not extensively analyzed. Furthermore, feature selection strategies are not discussed in detail.

For the present project, the key contribution of this study lies in demonstrating how **SVM performance can be improved through data reduction and weighting techniques**. Although the RW-SVM method itself is not directly implemented, the study highlights the importance of addressing class imbalance and considering computational efficiency when training SVM models.

### 2.3 Article 3 Analysis: Income Prediction Using Logistic Regression, Random Forest, and SVM

This study examines the effectiveness of three widely used machine learning algorithms—**Logistic Regression, Random Forest, and Support Vector Machines**—for predicting income levels using the Adult Income dataset. The dataset contains demographic and socioeconomic variables such as **age, education level, working hours, and marital status**.

The experimental design involves splitting the dataset into **70% training data and 30% testing data**, while also applying **5-fold cross-validation** during training to improve the reliability of the results. Logistic Regression is used as a baseline model due to its simplicity and interpretability, while Random Forest and SVM are applied to capture more complex relationships within the data. In particular, the SVM model is implemented using an **RBF kernel** to capture nonlinear decision boundaries.

The results indicate that all three models achieve strong predictive performance. The **SVM model reaches the highest classification accuracy (approximately 85.88%)**, while Logistic Regression obtains the **highest AUC score**, suggesting strong ranking capability. Random Forest also performs competitively while offering the additional advantage of feature importance estimation.

The feature analysis conducted in the study identifies several influential variables, including **marital status, age, and years of education**. These findings highlight the importance of demographic and socioeconomic characteristics in determining income levels.

However, the study provides limited information regarding preprocessing procedures. In particular, the handling of missing values and the encoding of categorical variables are not clearly described. Additionally, feature importance is primarily derived from the Random Forest model without comparison to alternative interpretability methods.

Despite these limitations, the study demonstrates that relatively simple machine learning models can achieve high predictive performance on the Adult Income dataset. It also reinforces the importance of evaluating models using **multiple performance metrics**, including accuracy, precision, recall, F1-score, and AUC.
